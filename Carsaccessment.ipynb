# Distribution of target and class

```python
plt.figure(figsize=(6,4))
sns.countplot(x="class", data=df1, order=df1["class"].value_counts().index, palette="Set2")
plt.title("Distribution of Car Acceptability")
plt.show()
```

# Safety vs class

```python
plt.figure(figsize=(6,4))
sns.countplot(x="safety", hue="class", data=df1, palette="Set1")
plt.title("Safety vs Car Acceptability")
plt.show()
```

# Buying price vs class

```python
plt.figure(figsize=(6,4))
sns.countplot(x="buying", hue="class", data=df1, palette="Set3")
plt.title("Buying Price vs Car Acceptability")
plt.show()
```

# Maintainance vs class

```python
plt.figure(figsize=(6,4))
sns.countplot(x="maint", hue="class", data=df1, palette="coolwarm")
plt.title("Maintenance vs Car Acceptability")
plt.show()
```
